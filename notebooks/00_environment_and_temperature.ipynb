# 00 — Environment and the temperature hurdle
Thermal occupation n̄ versus temperature for a 5 GHz microwave qubit and a 193 THz telecom photon, with lines at 15 mK and 300 K [clerk2010].

```
import math, numpy as np, matplotlib.pyplot as plt
from qll.circuits.noise.thermal import bose_einstein_occupation
T = np.logspace(-3, 3, 400)
for f, label in [(5e9, '5 GHz qubit'), (193.4e12, '193 THz photon')]:
    n = [max(bose_einstein_occupation(2*math.pi*f, t), 1e-40) for t in T]
    plt.loglog(T, n, label=label)
for t, name in [(0.015, '15 mK'), (300, '300 K')]:
    plt.axvline(t, ls='--', color='gray'); plt.text(t, 1e-30, name)
plt.xlabel('T (K)'); plt.ylabel('mean occupation n̄'); plt.legend(); plt.ylim(1e-40, 1e5); plt.show()
```
