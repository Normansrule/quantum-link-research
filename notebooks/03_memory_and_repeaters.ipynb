# 03 — Memories and repeaters: the thesis question in numbers

```
from qll.network.memory_decoherence import MEMORY_TABLE, BASELINES, capability_matrix, crossover_time_s
cm = capability_matrix(0.95)
for p in MEMORY_TABLE:
    print(f'{p.name:38s} crossover {crossover_time_s(0.95, p.lifetime_s, p.model)/60:9.1f} min  ' + ' '.join('Y' if cm[p.name][b] else '.' for b in BASELINES))
print('baselines (min):', {b: round(t/60,1) for b,t in BASELINES.items()})
```

## Purification: rounds, pairs, and classical round trips

```
from qll.network.purification import bbpssw_rounds_to_target, purify_werner
from qll.constants.astro import EARTH_MARS_MAX_M
for F0 in (0.6, 0.8, 0.9):
    rec = purify_werner(F0, 0.99, distance_m=EARTH_MARS_MAX_M)
    print(f'F0={F0}: {rec.rounds} rounds, {rec.expected_pairs_consumed:.0f} pairs, {rec.classical_time_s/3600:.1f} h of classical exchange at Mars max')
```

## Chain versus direct: the crossover moves with memory time

```
import numpy as np, matplotlib.pyplot as plt
from qll.network.repeater_chain import memory_chain, direct_rate_hz
L = np.logspace(1.5, 3.5, 200)
plt.loglog(L, [direct_rate_hz(x) for x in L], 'k', label='direct')
for T2 in (0.01, 1.0, 100.0):
    plt.loglog(L, [max(memory_chain(x, 3, T2).rate_hz, 1e-12) for x in L], label=f'8 segments, T2={T2} s')
plt.ylim(1e-6, 1e8); plt.xlabel('km'); plt.ylabel('pairs/s'); plt.legend(); plt.show()
```
