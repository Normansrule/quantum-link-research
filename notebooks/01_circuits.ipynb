# 01 — Circuits: teleportation fidelity vs stored delay, CHSH vs Werner fraction
Everything here calls the tested functions in `qll.circuits`; the analytic curves are the tests' expectations.

```
import numpy as np, matplotlib.pyplot as plt
from qll.circuits.bell import werner_state
from qll.circuits.teleportation import average_fidelity, analytic_average_fidelity
from qll.circuits.chsh import chsh_value, chsh_werner_analytic
fs = np.linspace(0.25, 1, 7)
F = [average_fidelity(werner_state(f), n_haar=100) for f in fs]
S = [chsh_value(werner_state(f)) for f in fs]
fig, (a, b) = plt.subplots(1, 2, figsize=(10, 3.6))
a.plot(fs, F, 'o', label='simulated'); a.plot(fs, [analytic_average_fidelity(f) for f in fs], label='(2f+1)/3'); a.axhline(2/3, ls=':', color='k'); a.set(xlabel='fully entangled fraction f', ylabel='teleportation fidelity'); a.legend()
b.plot(fs, S, 'o', label='computed'); b.plot(fs, [chsh_werner_analytic(f) for f in fs], label='2√2(4f−1)/3'); b.axhline(2, ls=':', color='k'); b.set(xlabel='f', ylabel='CHSH S'); b.legend()
plt.show()
```

## The light-time seal
A teleported qubit at Mars distance cannot be opened before the two bits arrive.

```
from qll.circuits.teleportation import teleport
from qll.channels.light_time_delay import NotYetArrived
from qll.constants.astro import EARTH_MARS_MIN_M
rec = teleport(np.array([0.6, 0.8]), distance_m=EARTH_MARS_MIN_M)
print('bits', rec.classical_bits, 'arrive after', rec.message.earliest_arrival_s/60, 'min')
try:
    rec.output.open(now_s=60.0)
except NotYetArrived as e:
    print('sealed:', e)
```
